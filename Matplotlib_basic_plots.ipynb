## Introduction to Matplotlib
Matplotlib is one of the most widely used libraries in the Python Programming Language for creating static, animated and interactive visualizations. 
Matplotlib is use as a data visualization tool in Data Science, Machine LEarning and Data Anlysis

### 1. what is Matplotlib?
Matplotlib is a Python plotting library that allows users to create vrious types of graphs including:
- Line Plots
- Bar charts
- Histograms
- scatter plots
- 3D plots
- Heatmaps
- Pie Charts
- Box plots

### 2. Why Use Matplotlib?
- <b>Versatile:</b> Matplotlib supports a variety of plot types that can be customized with colors, labels, grids, and legends.

- <b>Highly Customizable: </b> You can adjust nearly every aspect of your plot, from axes and labels to colors and line styles, making it suitable for both simple and complex visualizations.

- <b>Integration: </b> It integrates well with other scientific libraries like NumPy for numerical operations and Pandas for data analysis, providing an efficient workflow for data visualization.

- <b>Wide Adoption: </b> Matplotlib is one of the most popular plotting libraries in the Python ecosystem, meaning there are many resources and community support available.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
fs_continuous = 1000  # High sampling frequency for the continuous signal
f_signal = 5  # Frequency of the signal in Hz
T_signal = 1 / f_signal  # Period of the sig6y5utunal
t_continuous = np.linspace(0, 1, fs_continuous)  # Time vector for continuous signal
continuous_signal = np.sin(2 * np.pi * f_signal * t_continuous)  # Continuous sine wave

# Sampling frequencies b60 ,7[p78o8]
fs_high = 50  # Sampling rate greater than Nyquist rate (no aliasing)
jjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjjfs_low = 10  # Sampling rate below Nyquist rate (aliasing)

# Time vectors for sampled signals
t_high = np.arange(0, 1, 1/fs_high)
t_low = np.arange(0, 1, 1/fs_low)

# Sampled signals
high_sampled_signal = np.sin(2 * np.pi * f_signal * t_high)
low_sampled_signal = np.sin(2 * np.pi * f_signal * t_low)

# Plotting the continuous signal and sampled signals

plt.figure(figsize=(10, 8))

# Continuous signal plot
plt.subplot(3, 1, 1)
plt.plot(t_continuous, continuous_signal, label="Continuous Signal", color='blue')
plt.title("Continuous-Time Signal (5 Hz Sine Wave)")
plt.xlabel("Time [s]")
plt.ylabel("Amplitude")
plt.grid(True)

# High rate sampled signal (no aliasing)
plt.subplot(3, 1, 2)
plt.plot(t_continuous, continuous_signal, label="Continuous Signal", color='blue')
plt.stem(t_high, high_sampled_signal, linefmt='r-', markerfmt='ro', basefmt="r-", label="High Rate Samples")
plt.title("Signal Sampled at High Rate (50 Hz)")
plt.xlabel("Time [s]")
plt.ylabel("Amplitude")
plt.grid(True)

# Low rate sampled signal (aliasing)
plt.subplot(3, 1, 3)
plt.plot(t_continuous, continuous_signal, label="Continuous Signal", color='blue')
plt.stem(t_low, low_sampled_signal, linefmt='g-', markerfmt='go', basefmt="g-", label="Low Rate Samples")
plt.title("Signal Sampled at Low Rate (10 Hz) - Aliasing")
plt.xlabel("Time [s]")
plt.ylabel("Amplitude")
plt.grid(True)

# Display the plots
plt.tight_layout()
plt.show()
```
